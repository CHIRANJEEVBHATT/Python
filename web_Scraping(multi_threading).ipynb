```python
import threading
import requests
from bs4 import BeautifulSoup

urls=[
    'https://www.w3schools.com/python/',
'https://www.w3schools.com/programming/index.php',
'https://www.w3schools.com/aws/index.php'
]

def fetch_cintent(url):
  response=requests.get(url)
  soup=BeautifulSoup(response.content,'html.parser')
  title=soup.title.string
  print(f'Fetched {len(soup.text)} character from {url}')

threads= []

for url in urls:
  thread = threading.Thread(target=fetch_cintent,args=(url,))
  threads.append(thread)
  thread.start()

for thread in threads:
  thread.join()

  print("all pages are fetched")
```
