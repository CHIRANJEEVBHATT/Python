```python
df_filled = df.fillna(0)
```

```python
##filling missing value with mean of col
df['Sales_fillNA'] = df['Sales'].fillna(df['Sales'].mean())
```

```python
df
```

```python
df.dtypes
```

```python
#convert
df.rename(columns={'Sale Date':'Sales Date'})
df.head()
```

```python
#chnage datatype
df['Value_new'] = df['Value'].fillna(df['Value'].mean()).astype(int)
df.head()
```

```python
df['New Value'] = df['Value'].fillna(df['Value'].mean()).apply(lambda x:x*2).astype(int)
df.head()
```

```python
### data Aggregating and grouping
df.head()
```

```python
grouped_mean=df.groupby('Product')['Value'].mean()
print(grouped_mean)
```

```python
grouped_sum = df.groupby(['Product' , 'Region'])['Value'].sum()
```

```python
print(grouped_sum)
```

```python
##aggregiate multiple function
grouped_agg = df.groupby('Region')['Value'].agg(['mean','sum','count'])
print(grouped_agg)
```

```python
#merge and join dataframe
#create dataframes
df1 = pd.DataFrame({'key':['A','B','C'],'value':[1,2,3]})
df2 = pd.DataFrame({'key':['A','B','D'],'value':[3,4,5]})
```

```python
df1
```

```python
df2
```

```python
#merge on key column
df3 = pd.merge(df1,df2,on='key',how="inner")
print(df3)
```

```python
pd.merge(df1,df2,on='key',how="outer")
```

```python
pd.merge(df1,df2,on='key',how="left")
```

```python
pd.merge(df1,df2,on='key',how="right")
```
